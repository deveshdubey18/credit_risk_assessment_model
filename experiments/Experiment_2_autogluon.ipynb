```python
# importing libraryies
from autogluon.tabular import TabularPredictor
import pandas as pd
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.metrics import classification_report
```

```python
# loading data
df = pd.read_csv(r'C:\Devesh\Credit_Risk_Assessment_ML_Model\data\processed\processedData.csv')
```

```python
# splitting data into 2 parts (Training and Testing)
train_data,test_data = train_test_split(df,test_size=0.3,random_state=1)
```

```python
# using autogluon

predictor = TabularPredictor(label='is_at_risk').fit(train_data=train_data,time_limit=500)

prediction = predictor.predict(test_data.drop(columns='is_at_risk'))

print(prediction)
print(predictor.leaderboard(test_data))
print(predictor.model_best)
print(predictor.evaluate(test_data))


```

```python
report = classification_report(test_data['is_at_risk'],prediction)
print(report)
```
