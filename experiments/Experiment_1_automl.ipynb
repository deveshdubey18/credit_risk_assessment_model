## Model using AutoML (flaml)

```python
from flaml.automl.automl import AutoML
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.preprocessing import MinMaxScaler
from imblearn.over_sampling import SMOTE
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score,classification_report
```

```python
# data loading
df = pd.read_csv(r'C:\Devesh\Credit_Risk_Assessment_ML_Model\data\processed\processedData.csv')
```

```python
def preprocessing(df):
    
    X = df.drop(columns='is_at_risk')
    y = df['is_at_risk']
    
    df.drop_duplicates()
    X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=1)
    
    sc = MinMaxScaler()
    X_train = sc.fit_transform(X_train)
    X_test = sc.transform(X_test)
    
    smote = SMOTE()
    X_train,y_train = smote.fit_resample(X_train,y_train)
    
    return X_train,X_test,y_train,y_test
```

```python
X_train,X_test,y_train,y_test = preprocessing(df)
```

```python
automl = AutoML()
automl_setting = {'metric':'accuracy',
                  'task':'classification',
                  'time_budget':50}


automl.fit(X_train=X_train,y_train=y_train, **automl_setting)
print(automl.predict_proba(X_train))
# Print the best model
print(automl.model.estimator)
    
y_pred = automl.predict(X_test)
```

```python
# Evaluation
accuracy = accuracy_score(y_test,y_pred)
report = classification_report(y_test,y_pred)
    
print('Classification report :\n',report)
```

```python
automl = AutoML(time_budget=10,task='classification',metric='accuracy')
cv = cross_val_score(automl,X_train,y_train,cv=5)
print(np.mean(cv))
```
